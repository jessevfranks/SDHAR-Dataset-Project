## Task and Models

Our task is to focus on activity recognition, at least for now. We will create a few models to get a baseline

#### Random Forest
A random forest may be able to accurately predict activities, as this data is somewhat complex and noisy

#### KNN
K-Nearest Neighbors is likely not a good choice due to the curse of dimensionality

#### Logistic/Linear Regression
The relationships here aren't linear, so this likely wouldn't be a good choice

#### Decision Trees
These may be a good option due to the breadth of the data, but the data set is very large which may cause issues

#### DNN
Deep neural networks handle high-dimensional data well, but do not take into account sequencing. Due to this, DNNs are not a good choice

#### CNN
Convolutional neural networks are good for grid-like structures (images) and for spotting patterns in certain sectors of data. This may be a possible option, but likely wouldn't perform well. It's main use cases involve image, video, or natural language processing.

#### LSTM
Long Short-Tern Memory models are good for understanding sequential data, context, and long term dependencies. This is likely the best choice for the job (out of the deep models)

# Model analysis

### LSTM
The LSTM model has disappointing performance so far. While it has an overall accuracy of 92%, it completely fails to identify less frequent activities. There are some key improvements that can be made here, however. First, and likely most importantly, we can rebalance the activity distribution. Using techniques like SMOTE, we can "add" more of these activities to better train the model. We can also experiment with tuning the models hyperparameters and/or number of epochs. This should allow us to increase performance with a deep model

### Random Forest and Decision Tree
Both of these models performed similar. As we can see,  they had accuracies of 97.9% and 95.9% respectively. However, these numbers don't tell the whole truth. Because longer, more predictable activities like SLEEP, OUT HOME, and WATCH TV dominate the dataset, most of this performance comes from this. We can see that the models perform poorly on shorter, infrequent, less predicable activities like PET, TAKE MEDS, and DRESS. Random Forest performs definitively better than the Decision Tree. This was somewhat expected, and for the future, we will likely focus on using Random Forests for shallow models. Similarly to the LSTM model, the best thing we can do to increase performance is utilizing SMOTE to oversample minority activities. This should make a big difference in predicting these activities
