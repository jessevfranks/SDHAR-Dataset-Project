# Normalized Model analysis
We see that the Random Forest and Decision Tree models kept the same accuracy (+/- 0.1%). This is because Random Forests and Decision Trees do not care about scaled values. At least, they don't find linear relationships, so the effects of scaling are relatively minimal. When the random state is changed (ie from 42 to 60), the accuracies do differ, but this is also expected.

The LSTM model saw a roughly 0.8% jump in accuracy, from 92.02% to 92.81%. Although it does not seem like much, this is a relatively significant accuracy jump. We did not normalize that many columns, and we didn't have extreme variation, so it makes sense that normalization wouldn't bring a world of difference. Additionally, the dataset is still oversampled towards the more frequent activities. We will address this using SMOTE.
